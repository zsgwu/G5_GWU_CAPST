```python
import pandas as pd

instate_tuition = pd.read_csv('tuition-tracker-net.csv')
tuition = pd.read_csv('tuition-tracker-sticker.csv')
```

```python
instate_tuition.head()
```

```python
tuition.head()
```

```python
print('Missing values in instate_tuition:')
display(instate_tuition.isnull().sum())
```

```python
print('\nMissing values in tuition:')
display(tuition.isnull().sum())
```

```python
print('Rows with missing values in instate_tuition:')
display(instate_tuition[instate_tuition.isnull().any(axis=1)].head())
```

```python
print('\nRows with missing values in tuition:')
display(tuition[tuition.isnull().any(axis=1)].head())
```

```python
tuition = tuition.dropna()
print('Shape of tuition after dropping missing values:', tuition.shape)
```

```python
print('\nMissing values in tuition:')
display(tuition.isnull().sum())
```

```python
instate_tuition.to_csv('instate_tuition_cleaned.csv', index=False)
tuition.to_csv('tuition_cleaned.csv', index=False)
print('DataFrames saved to instate_tuition_cleaned.csv and tuition_cleaned.csv')
```
